```python
model = YOLO("best.pt")
import cv2

cap = cv2.VideoCapture(0) #open video

while True:#run forever
    ret, frame = cap.read() #frame by frame
    if not ret:
        break

    results = model(frame) # use my model to get prediction
    annotated_frame = results[0].plot()  # draw boxes

    cv2.imshow("YOLO Detection", annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
